# Colloquium Notes: Charlotte Ward

## Rubin + Euclid + Roman
- combining data from big surveys
- Rubin = deep optical, lots of galaxies/transients
- Euclid = higher res, optical + IR
- Roman = IR + spectroscopy/time domain
- use them together to get better measurements

## Blending problem
- lots of galaxies overlap in deep Rubin images
- ~30–60% could be blended
- makes galaxy shapes/brightness/redshifts harder
- Euclid higher res helps separate sources

## Scarlet 2
- code her group uses for deblending/joint modeling
- written in JAX, GPU compatible
- models different surveys at same time
- accounts for different pixel scales, PSFs, filters
- doesn't force galaxies into simple shapes

## Machine learning
- ML prior helps keep galaxy models realistic
- diffusion model trained on galaxy images
- mainly helps when sources overlap / fit becomes unstable
- prior guides fit, data still important

## Transients
- can model static + changing sources
- recover transient positions + light curves
- useful for supernovae
- can help when difference imaging has problems

## Lensed supernovae
- lensing can make multiple images of same SN
- different arrival times = time delays
- time delays can constrain Hubble constant
- tested if Rubin alone can recover them even when images are barely resolved

## Wandering black holes / TDEs
- some massive black holes may not be in galaxy centers after mergers
- hard to see if they aren't actively accreting
- TDE = star gets torn apart by black hole -> bright flare
- found TDEs offset from galaxy centers
- possible way to find wandering black holes

## Changing-look AGN
- broad emission lines can appear/disappear/change
- probably related to changes in accretion disk
- comparing observations to simulations

## Radio
- using ASKAP/VAST
- looking for AGN that suddenly get stronger in radio
- possible new jets
- some fade again -> probably radio transients, not long-term activity


## Questions
- How much will Euclid improve Rubin measurements of blended galaxies?
- How accurately can lensed-supernova time delays be recovered from Rubin data alone?
- How common are wandering massive black holes in low-mass galaxies?
